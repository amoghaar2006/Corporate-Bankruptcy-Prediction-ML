```python
from sklearn.model_selection import train_test_split

# Separate the input and target
X=df.drop(columns=["class"])
y=df["class"]


# Split into 80% training and 20% testing
X_train, X_test, y_train, y_test=train_test_split(X,y,test_size=0.20, random_state=42)

print("X_train:",X_train.shape)
print("y_train:",y_train.shape)
print("X_test:",X_test.shape)
print("y_test:",y_test.shape)
```

```python
from sklearn.impute import SimpleImputer
import numpy as np

#Create a mean imputer
imputer=SimpleImputer(strategy="mean")

#Calculate feature means from the training data and replace missing values in X_train

X_train=imputer.fit_transform(X_train)

#Same for the test data

X_test=imputer.transform(X_test)

print("Missinng values in X_train:",np.isnan(X_train).sum())
print("Missing values in X_test",np.isnan(X_test).sum())
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model=LogisticRegression(max_iter=1000)

#Train the model
logistic_model.fit(X_train, y_train)
print("Logistic Regression model trained successfully")
```

```python
y_pred=logistic_model.predict(X_test)

print("First 20 predictions")
print(y_pred[:20])
```

```python
import time
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, log_loss, confusion_matrix
logistic_model=LogisticRegression(solver="liblinear", max_iter=5000)

#REcord the starting time
start_time=time.time()

logistic_model.fit(X_train, y_train)

#Calculating how long training took
training_time=time.time()-start_time

# Predict classes for the training data
y_train_pred=logistic_model.predict(X_train)
# Predict classes for the test data
y_test_pred=logistic_model.predict(X_test)

# Get prediction probabilties for calculating log loss
y_test_prob=logistic_model.predict_proba(X_test)

#Calculate the training accruacy
train_accuracy=accuracy_score(y_train,y_train_pred)


# Calculate test accuracy
test_accuracy = accuracy_score(y_test, y_test_pred)

#Calculate the test log loss
test_log_loss=log_loss(y_test, y_test_prob)

#Calculate the confusion matrix

cm=confusion_matrix(y_test, y_test_pred)

# Display results
print("Logistic Regression Results")
print("---------------------------")
print("Training Accuracy :", train_accuracy)
print("Test Accuracy     :", test_accuracy)
print("Test Log Loss     :", test_log_loss)
print("Training Time     :", training_time, "seconds")

print("\nConfusion Matrix:")
print(cm)
```
